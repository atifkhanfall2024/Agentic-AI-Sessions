```python
from dotenv import load_dotenv
import os

load_dotenv()

os.environ['GROQ_API_KEY'] = os.getenv('GROQ_API_KEY')
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model='llama-3.3-70b-versatile')
```

```python
from pydantic import BaseModel , Field

class Actor(BaseModel):
    name :str = Field(... , description='Name of Actor')
    role:str = Field(... , description='Role of Actors')

class MovieDetail(BaseModel):
    title:str = Field(...,description='Name of Movie')
    director:str = Field(... , description='The director of movie')
    rating:float  
    cast:list[Actor]
    year :int =Field(description='Year of Release')
    budget:int | None = Field(description='Budget of Movie in PKR')
```

```python
structured_output = model.with_structured_output(MovieDetail)
response = structured_output.invoke("Hi famous movie of amir khan ? ")
print(response)
```

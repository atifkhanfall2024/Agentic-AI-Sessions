## TypeDict
 - used to define the structure output of dictionary

```python
from dotenv import load_dotenv
import os

load_dotenv()

os.environ['GROQ_API_KEY'] = os.getenv('GROQ_API_KEY')
```

```python
from langchain.agents import create_agent
from dataclasses import dataclass
from langchain_groq import ChatGroq


@dataclass
class UserInfo:
    name:str 
    email:str
    age:int
    height:float

models = ChatGroq(
    model='llama-3.3-70b-versatile'
    
    )

agent = create_agent(
    model=models ,
    response_format=UserInfo
)


response = agent.invoke({"messages" :[{"role":"user" , "content":"Hi give me the detail of shahrukh"}]})

print(response['structured_response'])
```

```python
from langchain.agents import create_agent
from typing_extensions import TypedDict
from langchain_groq import ChatGroq



class UserInfo(TypedDict):
    name:str 
    email:str
    age:int
    height:float

models = ChatGroq(
    model='llama-3.3-70b-versatile'
    
    )

agent = create_agent(
    model=models ,
    response_format=UserInfo
)


response = agent.invoke({"messages" :[{"role":"user" , "content":"Hi give me the detail of shahrukh"}]})

print(response['structured_response'])
```

```python
from langchain.agents import create_agent
from pydantic import BaseModel , Field
from langchain_groq import ChatGroq



class UserInfo(BaseModel):
    name:str=Field(... , description='Name of User')
    email:str=Field(... ,description= 'Email of user')
    age:int=Field(... , description='Age of user')
    height:float=Field(... , description='height of user')

models = ChatGroq(
    model='llama-3.3-70b-versatile'
    
    )

agent = create_agent(
    model=models ,
    response_format=UserInfo
)


response = agent.invoke({"messages" :[{"role":"user" , "content":"Hi give me the detail of shahrukh"}]})

print(response['structured_response'])
```

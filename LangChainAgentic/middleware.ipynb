## MIDDLEWARE
- middle ware is used to get tightly more control inside the agent

```python
import os
from dotenv import load_dotenv

load_dotenv()
os.environ['GROQ_API_KEY'] = os.getenv('GROQ_API_KEY')
```

```python
from langchain.agents import create_agent
from langchain_groq import ChatGroq
from langchain.agents.middleware import SummarizationMiddleware
from langgraph.checkpoint.memory import InMemorySaver
from langchain_core.messages import HumanMessage , AIMessage , SystemMessage

models = ChatGroq(model='llama-3.3-70b-versatile')

agent = create_agent(
    model=models,
    checkpointer=InMemorySaver(),
    middleware=[SummarizationMiddleware(
        model=models,
        keep=('messages' , 4),
        trigger=('messages' , 10)
    )
    ]

)
```

```python
config = {"configurable":{"thread_id":"atifkhan"}}
```

```python
questions = {
    "what is 4*4" , 
    "what is 2 +2" ,
    "what is 2+2-2" ,
    "what is 22222+111111" ,
    "what is 2345678-123456789"
}

for q in questions:
    response = agent.invoke({"messages":[HumanMessage(content=q)]} , config)
    print(f"Messages : {response} ")
    print(f"{len(response['messages'])}")
```
